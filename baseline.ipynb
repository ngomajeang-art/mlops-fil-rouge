```python
# --- 1. Importer les outils dont on a besoin ---
from sklearn.datasets import load_diabetes # le jeu de données
from sklearn.model_selection import train_test_split # séparer les données
from sklearn.ensemble import RandomForestRegressor # le modèle (forêt d'arbres)
from sklearn.metrics import mean_absolute_error, r2_score # mesurer la qualité

# --- 2. Charger les données ---
# X = les caractéristiques des patients (âge, IMC, tension...)
# y = ce qu'on veut prédire (l'évolution de la maladie après un an)

data = load_diabetes(as_frame=True)

X, y = data.data, data.target

# --- 3. Séparer en deux parties ---
# 80 % pour apprendre (train), 20 % pour tester (test) sur des données jamais vues.
# Pas de graine aléatoire : le découpage change à chaque exécution (volontaire).
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
# --- 4. Créer et entraîner le modèle ---
# La forêt aléatoire combine 100 arbres de décision. Chaque arbre donne son avis,
# et la prédiction finale est la moyenne de ces avis.

model = RandomForestRegressor(n_estimators=100)
# « fit » = le modèle apprend à partir des données d'entraînement
model.fit(X_train, y_train)

# --- 5. Faire des prédictions sur les données de test ---
pred = model.predict(X_test)

# --- 6. Mesurer la qualité ---
# MAE : erreur moyenne entre prédiction et réalité (plus c'est petit, mieux c'est)
# R2 : part de la variation expliquée par le modèle (1 = parfait, 0 = inutile)
print("MAE :", mean_absolute_error(y_test, pred))
print("R2 :", r2_score(y_test, pred))
```
